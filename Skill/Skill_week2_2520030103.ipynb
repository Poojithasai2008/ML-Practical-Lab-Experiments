```python
import pandas as pd
import zipfile
import os

# File name
CSV_PATH = "amazon.csv"
ZIP_FILE_PATH = "amazon.csv.zip"


if os.path.exists(ZIP_FILE_PATH):
    with zipfile.ZipFile(ZIP_FILE_PATH, 'r') as zip_ref:
        zip_ref.extractall('.') # Extract to current directory
    print(f"'{ZIP_FILE_PATH}' unzipped successfully.")

# Read CSV file
df = pd.read_csv(CSV_PATH)

# 1. Display first 5 rows
print("FIRST 5 ROWS")
print(df.head())

# 2. Display dataset shape
print("\nDATASET SHAPE")
print(df.shape)

# 3. Display column names
print("\nCOLUMN NAMES")
print(df.columns)

# 4. Display data types
print("\nDATA TYPES")
print(df.dtypes)

# 5. Display summary statistics
print("\nSUMMARY STATISTICS")
print(df.describe())

# 6. Display information about dataset
print("\nDATASET INFORMATION")
df.info()

# 7. Display missing values
print("\nMISSING VALUES")
print(df.isnull().sum())

# 8. Display duplicate values
print("\nDUPLICATE VALUES")
print(df.duplicated().sum())

# 9. Remove duplicate rows
df = df.drop_duplicates()

# 10. Display unique values in each column
print("\nUNIQUE VALUES IN EACH COLUMN")
for column in df.columns:
    print(f"{column}: {df[column].nunique()}")

# 11. Display total number of rows and columns
print("\nROWS AND COLUMNS")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

# 12. Display percentage of missing values
print("\nPERCENTAGE OF MISSING VALUES")
missing_percentage = (df.isnull().sum() / len(df)) * 100
print(missing_percentage)

# 13. Display percentage of duplicate rows
print("\nPERCENTAGE OF DUPLICATE ROWS")
duplicate_percentage = (df.duplicated().sum() / len(df)) * 100
print(duplicate_percentage)

# 14. Display sample data
print("\nRANDOM SAMPLE OF 5 ROWS")
print(df.sample(5))
```
